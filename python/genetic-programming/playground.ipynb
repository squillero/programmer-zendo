```python
import operator
import gxgp
```

```python
gp = gxgp.DagGP(operators=[operator.add, operator.sub, operator.mul],
                variables=2, constants=5)
```

```python
i1 = gp.create_individual(10)
i2 = gp.create_individual(10)
```

```python
i1.draw()
```

```python
i2.draw()
```

```python
o = gxgp.xover_swap_subtree(i1, i2)
o.draw()
```
